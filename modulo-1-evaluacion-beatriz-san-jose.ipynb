## 2. Funciones

### 1. Agregar producto al inventario

```python
# Revisar la lista para ver si existe o no: for
# Si existe, actualizar la cantidad: suma
# Si no existe, agregar producto: .append


def agregar_producto (nombre, precio, cantidad):
    for producto in inventario:
        if producto ["nombre"].lower() == nombre.lower():
            cantidad_aniadida = cantidad
            producto ["cantidad"] += cantidad_aniadida
            cantidad_actualizada = cantidad_aniadida + producto ['cantidad']
            print (f"La cantidad de {nombre} se ha actualizado, ahora es {producto ['cantidad']}")
            return inventario
    else:
        
        inventario.append({'nombre': nombre, 'precio': precio, 'cantidad': cantidad})
        print (f"{nombre} ha sido añadido al inventario con éxito")
        return inventario
            

```

### 4. Actualizar stock

```python
# Saber si el producto (nombre del producto) está en la lista: bucle for
# Si está, actualizar la cantidad + return
# Si no está, imprimir mensaje

def actualizar_stock (nombre, cantidad):
    for producto in inventario:
        if producto ["nombre"].lower() == nombre.lower():
            cantidad_aniadida = cantidad
            producto ["cantidad"] += cantidad_aniadida
            cantidad_actualizada = cantidad_aniadida + producto ['cantidad']
            print (f"El stock de {nombre} se ha actualizado, ahora es de {producto ['cantidad']} unidades")
            return inventario
    else:
        print (f"El producto {nombre} no está en el inventario")
```

```python
actualizar_stock ("agenda", 10)
```
