### 1. Agregar producto al inventario

```python
# Revisar la lista para ver si existe o no: for
# Si existe, actualizar la cantidad: suma
# Si no existe, agregar producto: .append


def agregar_producto (nombre, precio, cantidad):
    for producto in inventario:
        if producto ["nombre"].lower() == nombre.lower():
            cantidad_aniadida = cantidad
            producto ["cantidad"] += cantidad_aniadida
            cantidad_actualizada = cantidad_aniadida + producto ['cantidad']
            print (f"La cantidad de {nombre} se ha actualizado, ahora es {producto ['cantidad']}")
            return inventario
    else:
        
        inventario.append({'nombre': nombre, 'precio': precio, 'cantidad': cantidad})
        print (f"{nombre} ha sido añadido al inventario con éxito")
        return inventario
            

```

```python
# Saber si el producto (nombre del producto) está en la lista: bucle for
# Si está, actualizar la cantidad + return
# Si no está, imprimir mensaje

def actualizar_stock (nombre, cantidad):
    for producto in inventario:
        if producto ["nombre"].lower() == nombre.lower():
            cantidad_aniadida = cantidad
            producto ["cantidad"] += cantidad_aniadida
            cantidad_actualizada = cantidad_aniadida + producto ['cantidad']
            print (f"El stock de {nombre} se ha actualizado, ahora es de {producto ['cantidad']} unidades")
            return inventario
    else:
        print (f"El producto {nombre} no está en el inventario")
```

### 5. Eliminar un producto del inventario


```python
# 1. Tengo que saber si ese producto está o no en el inventario para poder eliminarlo
# 2. Usar bucle for y buscar por nombre
# 3. Si existe, eliminar el producto .pop()
# 4. Si no, print mensaje indicándolo

def eliminar_producto (nombre):
    for producto in inventario:
        if producto ["nombre"].lower() in inventario:
            producto ['nombre'].pop()
            print (f"El producto {nombre} se ha eliminado")
            return inventario
            
    else:
        print (f"El producto {nombre} no está en el inventario")
```

### 6. Calcular el valor del inventario

```python
# Crear bucle para recorrer la lista
# Multiplicar la cantidad por las uds de cada producto
# Sumar todos los resultados

def calcular_valor_inventario():
    valor_total = 0
    for producto in inventario:
        valor_producto = producto ["precio"] * producto ["cantidad"]
        valor_total += valor_producto
    print (f"El valor total del inventario es {valor_total}")
    return valor_total
```

### 7. Realizar una compra

```python

# Mostrar inventario
#Seleccionar un producto y agregarlo al carrito
#Actualizar el inventario
#Saber si quiere seguir comprando y repetir el proceso
# Input: nombre del producto que el cliente quiere comprar
# Agrupar los productos (nombre, cantidad) seleccionados y actualizar el inventario (restarlo del inventario)
# calcular el coste total

def realizar_compra():
    carrito = []
    compra_final = 0
    aniadir_carrito = "sí"
    while aniadir_carrito == "sí" or aniadir_carrito =="si":
        print (inventario)
        producto_elegido = input ("¿Qué producto desea agregar al carrito?").lower()
        cantidad_producto = int (input("¿Cuántas unidades desea agregar al carrito"))
        for producto in inventario:
            if producto_elegido == producto ["nombre"].lower():
                if producto ["cantidad"] >= cantidad_producto:
                    carrito_precio = producto ["precio"] * cantidad_producto
                    compra_final += carrito_precio
                    carrito.append({"nombre": producto ["nombre"], "cantidad" : cantidad_producto, "precio": carrito_precio})
                    producto ["cantidad"] = producto ["cantidad"] - cantidad_producto
                    print (f"Resumen compra: {carrito}")
                    aniadir_carrito = input ("¿Desea seguir comprando? Indique sí o no").lower() 
                    if aniadir_carrito == "no":
                        print (f'El precio final es {compra_final}')
                        break
        

                elif producto ["cantidad"] < cantidad_producto:
                    print (f"Solo quedan {producto ['cantidad']} unidades")
                    break
        else:
            print ("El producto indicado no se encuentra disponible")
    print ("Proceso de compra finalizado. El inventario queda así:")
    for producto in inventario:
        print(f"{producto['nombre']}: {producto ['cantidad']} unidades")      
    
    
    
```
